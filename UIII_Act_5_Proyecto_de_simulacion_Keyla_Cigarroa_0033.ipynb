```python
# Simulación Interactiva: El Carpintero Inteligente
# Reglas: Solo condicionales y operadores. Sin ciclos, sin funciones, sin POO.

print("====================================================")
print(" ¡BIENVENIDO AL TALLER DE CARPINTERÍA INTELIGENTE! ")
print("====================================================")
print("Eres el asistente del maestro carpintero.")
print("Ayúdanos a decidir qué hacer con la madera de hoy.\n")

# --- PASO 1: Selección de la Madera ---
print("¿Qué tipo de madera tienes disponible hoy?")
print("1. Roble (Madera dura, resistente)")
print("2. Pino (Madera blanda, fácil de trabajar)")
print("3. Caoba (Madera de lujo, costosa)")
tipo_madera = input("Introduce el número de tu opción (1, 2 o 3): ")

# --- PASO 2: Evaluación del Estado de la Madera ---
print("\n¿En qué estado se encuentra la madera?")
humedad = float(input("Introduce el porcentaje de humedad de la madera (ej. 12.5): "))
tiene_nudos = input("¿La madera tiene nudos que afecten su estructura? (si/no): ").lower()

# --- PASO 3: El Carpintero Artificial Toma Decisiones (Lógica Condicional) ---

print("\n====================================================")
print("       DIAGNÓSTICO DEL CARPINTERO INTELIGENTE       ")
print("====================================================")

# Caso crítico: Madera demasiado húmeda (No se puede trabajar)
if humedad > 20.0:
    print("Resultado: La madera está demasiado verde o húmeda (+" + str(humedad) + "%).")
    print("Acción: El carpintero decide mandarla al horno de secado. No se puede fabricar nada hoy.")

# Caso: Madera en buen estado
elif humedad <= 20.0:
    print("Estado de humedad: Óptimo para trabajar (" + str(humedad) + "%).")

    # Evaluación para ROBLE
    if tipo_madera == "1":
        print("Material: Roble detectado.")
        if tiene_nudos == "si":
            print("Análisis: Al ser roble con nudos, es difícil de tallar pero ideal para un estilo rústico.")
            print("Proyecto recomendado: Diseñar una mesa de taberna rústica.")
        else:
            print("Análisis: Roble de alta calidad y limpio.")
            print("Proyecto recomendado: Construir las estructuras de soporte para una casa o suelo de alta resistencia.")

    # Evaluación para PINO
    elif tipo_madera == "2":
        print("Material: Pino detectado.")
        # Uso de operadores lógicos (and / or)
        if tiene_nudos == "si" or humedad > 15.0:
            print("Análisis: Pino con imperfecciones o humedad ligeramente alta.")
            print("Proyecto recomendado: Construir cajas de embalaje o listones ocultos para construcción.")
        else:
            print("Análisis: Pino limpio y seco. Excelente para muebles de interior.")
            print("Proyecto recomendado: Fabricar una estantería ligera o un escritorio juvenil.")

    # Evaluación para CAOBA
    elif tipo_madera == "3":
        print("Material: Caoba detectada.")
        if tiene_nudos == "si":
            print("Análisis: ¡Alerta! Caoba con nudos. Es un desperdicio para muebles finos.")
            print("Proyecto recomendado: Tallar pequeñas artesanías o detalles decorativos para rescatar las partes limpias.")
        else:
            print("Análisis: Caoba perfecta, lisa y seca.")
            print("Proyecto recomendado: Fabricar un mueble de lujo para televisión o una guitarra acústica premium.")

    # Si introducen una opción de madera inválida
    else:
        print("Error: El carpintero no reconoce ese tipo de madera. ¡Asegúrate de marcar 1, 2 o 3!")

else:
    print("Error: Datos de humedad no válidos.")

print("\n====================================================")
print("          FIN DE LA SIMULACIÓN DEL TALLER           ")
print("====================================================")

print("Programa Realizado por Keyla Paola Cigarroa Palacios NL = 12")
```
